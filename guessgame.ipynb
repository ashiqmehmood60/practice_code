```python
!pip install streamlit

import random
import streamlit as st

st.set_page_config(
    page_title="Guess the Number",
    page_icon="🎯"
)

st.title("🎯 Number Guessing Game")

st.write("Guess a number between 1 and 100!")

if "secret_number" not in st.session_state:
    st.session_state.secret_number = random.randint(1, 100)
    st.session_state.attempts = 0
    st.session_state.game_over = False
    st.session_state.message = "Enter your first guess!"

guess = st.number_input(
    "Your guess",
    min_value=1,
    max_value=100,
    step=1,
    value=50,
    disabled=st.session_state.game_over
)

if st.button("Check Guess", disabled=st.session_state.game_over):

    st.session_state.attempts += 1

    if guess < st.session_state.secret_number:
        st.session_state.message = "Too low! Try bigger."

    elif guess > st.session_state.secret_number:
        st.session_state.message = "Too high! Try smaller."

    else:
        st.session_state.message = (
            f"Congratulations! You won in "
            f"{st.session_state.attempts} attempts!"
        )
        st.session_state.game_over = True

st.info(st.session_state.message)

st.write("Attempts:", st.session_state.attempts)

if st.session_state.game_over:

    st.success("You won the game!")

    if st.button("Play Again"):
        st.session_state.secret_number = random.randint(1, 100)
        st.session_state.attempts = 0
        st.session_state.game_over = False
        st.session_state.message = "Enter your first guess!"
        st.rerun()
```
